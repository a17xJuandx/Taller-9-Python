#### a) Diseña un programa que lea una **cadena** y un número entero $k$ y nos diga si alguna de sus palabras ingresadas tiene una longitud de $k$ caracteres.

```python
# Aquí debajo escribe tu solución.
cadena = "Hola mundo"
k = 3
a = cadena.split(" ")
for i in range(len(a)):
    if len(a[i]) == k:
        print(f"La palabra {a[i]} tiene una logintud de {k}")
    else:
        print(f"La palabra {a[i]} no tiene una logintud de {k}")
```

#### c) Convierte la siguiente **cadena** de caracteres en minúsculas y mayúsculas, con dos versiones: bucles + condicionales y otra con funciones `upper()` y `lower()`

*"La energía eléctrica es la forma de energía que resulta de la existencia de una diferencia de potencial entre dos puntos, lo que permite establecer una corriente eléctrica."*

* Puedes usar string.ascii_lowercase y string.ascii_uppercase.

```python
# Aquí debajo escribe tu solución.
cadena = "La energía eléctrica es la forma de energía que resulta de la existencia de una diferencia de potencial entre dos puntos, lo que permite establecer una corriente eléctrica."

a = cadena.upper()
b = cadena.lower()
print(a)
print(b)

import string


mayusculas = ""
minusculas = ""

for c in cadena:
    if c in string.ascii_lowercase:
        pos = string.ascii_lowercase.index(c)
        mayusculas += string.ascii_uppercase[pos]
    else:
        mayusculas += c

for c in cadena:
    if c in string.ascii_uppercase:
        pos = string.ascii_uppercase.index(c)
        minusculas += string.ascii_lowercase[pos]
    else:
        minusculas += c

print("\nVersión en MAYÚSCULAS (bucle):")
print(mayusculas)
print("\nVersión en minúsculas (bucle):")
print(minusculas)
```

#### d) Diseña un programa que lea una **lista** de 10 enteros, pero asegurándose de que todos los números introducidos por el usuario son positivos. Cuando un número sea negativo, lo indicaremos con un mensaje y permitiremos al usuario repetir el intento cuantas veces sea preciso.

#### k) Escribe un programa que verifique si una **matriz cuadrada** es **simétrica** (`A == AT`).

```python
# Aquí debajo escribe tu solución.
r = True
A = [[4,2,3],[3,2,3],[4,5,6]]
At = [[A[i][j] for i in range (len(A))] for j in range (len(A[0]))]
for i in range (len(A)):
  for j in range (len(At)):
    print(f"El numero {A[i][j]} es igual a {At[j][i]}? {A[i][j]==At[j][i]}")
    if A[i][j] != At[j][i]:
      r = False
if r == False:
  print("La matriz cuadrada no es simetrica")
else:
  print("La matriz cuadrada es simetrica")
```
